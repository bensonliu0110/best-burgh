**Dataset:** Pittsburgh SNAP Census Data - Public Safety 2010 (WPRDC)

**Link:** https://data.wprdc.org/dataset/pgh/resource/bd152875-a14d-421a-8656-cc896a7e9074

```python
import pandas as pd
import matplotlib.pyplot as plt

# load crime data
safety_df = pd.read_csv("safety.csv")

# preview the data
print(safety_df.head(10))
```

## The Metric

To measure safety, I used the Public Safety 2010 from the Pittsburgh SNAP Census Data, whichs tracks crime reports by neighborhood, including murder, rape, robbery, aggravated assault, burglary, and auto theft. My sub-metric is **Part 1 Crime per 100 Persons**, which measures the rate of crimes in relative to the neighborhood's population. I decided to use the rate instead of raw crime counts because large neighborhoods tend to have more people, which makes the number of crime reports higher than a smaller neighborhood. This makes an unfair comparison because we don't know if the numerous reports in large neighborhood fall under the category of serious crimes. By using rates, we can better compare the two and understand which neighborhood is generally more safer when considering the number of these crimes being reported. A lower rate means a safer neighborhood, so I ranked all neighborhoods from lowest to highest. 

**Dataset:** Pittsburgh SNAP Census Data - Public Safety 2010 (WPRDC)

**Link:** https://data.wprdc.org/dataset/pgh/resource/bd152875-a14d-421a-8656-cc896a7e9074

```python
top_10 = safety_df.sort_values("Part 1 Crime per 100 Persons (2010)").head(10)

plt.figure(figsize=(10, 5))
plt.barh(top_10['Neighborhood'], top_10['Part 1 Crime per 100 Persons (2010)'], color='skyblue')
plt.xlabel('Part 1 Crimes per 100 Persons')
plt.ylabel('Neighborhood')
plt.title('Top 10 Safest Neighborhoods in Pittsburgh')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

## The Best Neighborhood

Based on the Part 1 crime rate, the safest neighborhood is **Fairywood**, with just about 0.3 Part 1 crimes per 100 persons. It is followed by neighborhoods like New Homestead (0.7), Lincoln Place (1.0), and Summer Hill (1.2). Based on the bar chart, the pattern lies in the fact that smaller neighborhoods with less crime rates tend to be the safest option, as Fairywood met both requirements in the data. In contrast, the busier and larger neighborhoods tend to have much higher rates, and our only inference would be that there are more people living there, which makes crimes more common to occur.
