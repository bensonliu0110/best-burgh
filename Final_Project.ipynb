```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("safety_data.csv", low_memory=False)

df.columns = df.columns.str.upper()

clean_df = df.dropna(subset=['NEIGHBORHOOD'])

safety_df = clean_df.groupby('NEIGHBORHOOD').size().reset_index(name='total_incidents')

safety_df = safety_df.sort_values(by='total_incidents', ascending=True)

print(safety_df.head(10))
```

```python
top_10 = safety_df.head(10)

plt.figure(figsize=(10, 5))
plt.barh(top_10['NEIGHBORHOOD'], top_10['total_incidents'], color='skyblue')
plt.xlabel('Total Service Requests / Incidents')
plt.ylabel('Neighborhood')
plt.title('Top 10 Safest Neighborhoods in Pittsburgh')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

To measure neighborhood safety and general quality of living, we analyzed the total number of 311 service requests and incident reports in Pittsburgh. In this sub-metric, a lower number of reported incidents corresponds to a higher level of safety and community stability.

Southside Slopes ranked as the safest neighborhood according to our dataset, recording only 14 total incidents. Southside Flats followed closely behind with 43 incidents. Surrounding residential block groups and areas like Arlington Heights with 233 incidents and Glen Hazel with 606 incidents also placed in the top 10 safest areas.   

Based strictly on 311 incident volume, residential areas on the South Side and smaller residential pockets demonstrate significantly lower incident rates compared to dense commercial districts.
