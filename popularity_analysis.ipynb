## Introduction

Our group project is about finding the "best" neighborhood in Pittsburgh. We defined "bestness" through three sub-metrics: popularity, safety, and value. This notebook specifically covers the popularity aspect of the topic. 

**My Approach:** I believe that a neighborhood is considered popular if a lot of people are actively living there. The most straightforward way to measure this is finding the raw number of occupied housing units. 

**Alternatives I explored:** I also considered using total population, but that could include people who didn't live there. As a result, it could impact the results because it doesn't capture the actual number of people living in a specific neighborhood. Another alternative I considered was using the percentages of units occupied, but this would create a bias that favors smaller neighborhoods that happen to be full. Together, I believe that having the raw number of occupied units is best for measuring the popularity of a neighborhood because it tracks where the most households are actively choosing to live.

**Dataset:** Pittsburgh SNAP Census Data - Housing 2010 (WPRDC)

**Link:** https://data.wprdc.org/dataset/pgh/resource/94f8eda2-fa77-49a2-9190-3a6ed85fc561

```python
import pandas as pd

# Load the dataset from the file named 'housing.csv'
df = pd.read_csv("Popularity-Sub-Metric/housing.csv")

# prints out the first 5 rows to ensure dataset is loaded properly
print(df.head())
```

```python
# keeps three (neighborhood, population, and occupied units) out of 29 columns from the dataset
pop_data = df[['Neighborhood', 'Population (2010)', '# Occupied Units (2010)']]

# prints out the first five rows of the filtered data to ensure the selected columns are displayed
print(pop_data.head())
```

## The Metric

I believe that my metric follows a simple formula, where **Popularity** is based on **the number of occupied units** in 2010.

By filtering my dataset to three columns, the occupied units specifically tracks the number of housing units that are occupied. This could then be inferred that the more households that are occupied in a neighborhood, the more popular it is because of how many people are actively living there.

I have also included the **Population (2010)** column because it tells us how many peole are living in a neighborhood. While a popular neighborhood should have a high number of occupied units, I believe that it should also have a high population count. Together, the two columns is what helps me determine the most popular neighborhood in Pittsburgh.

**Dataset:** Pittsburgh SNAP Census Data - Housing 2010 (WPRDC)

**Link:** https://data.wprdc.org/dataset/pgh/resource/94f8eda2-fa77-49a2-9190-3a6ed85fc561

```python
# sorts the dataset in descending order based on the number of occupied units
sorted_data = pop_data.sort_values('# Occupied Units (2010)', ascending=False)

# prints out the top 10 neighborhoods
top_10_neighborhoods = sorted_data.head(10)
print(top_10_neighborhoods)
```

```python
# creates a bar chart that displays the top 10 neighborhoods by occupied units in 2010
top_10_neighborhoods.plot.bar(x='Neighborhood', y='# Occupied Units (2010)', title='Top 10 Neighborhoods by Occupied Units (2010)')
```

## The Best Neighborhood

Based on my popularity sub-metric, **Shadyside** is the best neighborhood. It has the most number of occupied housing units in Pittsburgh (8185), which means that most households have actively chose to live there. In addition, it also has a very high population (13915), which confirms the fact that there are a lot of people actually living in Shadyside and makes it a popular neighborhood in Pittsburgh.
